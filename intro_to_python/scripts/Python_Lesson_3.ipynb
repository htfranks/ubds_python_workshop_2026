 methods go after the variables and use a . and empty parentheses

```python
my_string = 'Hello world!' # create a string

print(len(my_string))

print(my_string.swapcase()) # Calling the swapcase method of the my_string variable

print(my_string.__len__()) # calling the internal len method
```

```python
print(my_string.isupper()) # method to ask if this string is uppercase
print(my_string[0].isupper())
print(my_string.upper().isupper())
```
