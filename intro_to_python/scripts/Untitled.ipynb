```python
# How to use pandas plotting
data = pd.read_csv('../data/gapminder_gdp_oceania.csv', index_col='country')

years = data.columns.str.replace('gdpPercap_','')
data.columns = years.astype(int)

data.loc['Australia'].plot()
```

```python
data.plot() # pandas plotting plots rows as default
```

```python
data.T.plot()
```

```python
data.T.plot(kind='bar')
plt.ylabel('GDP per capita')
```

# Matplotlib Plotting

```python
years = data.columns
gdp_australia = data.loc['Australia']

plt.plot(years, gdp_australia)
```

```python
plt.plot(years,gdp_australia, 'r--')
```

```python
gdp_australia = data.loc['Australia']
gdp_nz = data.loc['New Zealand']

# Plot our two countries 
plt.plot(years, gdp_australia, 'k-', label='Australia')
plt.plot(years, gdp_nz, 'r-', label='New Zealand') #hold is implied in Python 

# Create legend
plt.legend(loc = 'upper left')
plt.xlabel('Year')
plt.ylabel('GDP per capita ($)')
```

```python
# help(plt)

plt.scatter(gdp_australia,gdp_nz)
```

```python
# Can also do this in Pandas

data.T.plot.scatter(x='Australia', y = 'New Zealand')
```

```python
# Making a plot with min and max for Europe

data_europe = pd.read_csv('../data/gapminder_gdp_europe.csv', index_col='country')

data_europe.min().plot(label='min', color='r')
data_europe.max().plot(label = 'max', color='k')
plt.xlabel('GDP per capita ($)')
plt.xticks(rotation=40)
```
